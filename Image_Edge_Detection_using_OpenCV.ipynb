### Image Edge Detection using OpenCV

This notebook demonstrates how to perform edge detection using the **Canny Edge Detection** algorithm in OpenCV.

We will:
1. Create a synthetic image to ensure the code runs out-of-the-box.
2. Apply the Canny Edge Detector (`cv2.Canny`).
3. Display the original image alongside the detected edges.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Create a synthetic image (a white square on a black background)
# You can replace this with your own image using: img = cv2.imread('path_to_image.jpg', cv2.IMREAD_GRAYSCALE)
img = np.zeros((300, 300), dtype=np.uint8)
cv2.rectangle(img, (50, 50), (250, 250), 255, -1)

# 2. Apply Canny Edge Detection
# The two numbers (100, 200) are the minVal and maxVal thresholds
edges = cv2.Canny(img, threshold1=100, threshold2=200)

# 3. Plot and compare the results
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

axes[0].imshow(img, cmap='gray')
axes[0].set_title('Original Image')
axes[0].axis('off')

axes[1].imshow(edges, cmap='gray')
axes[1].set_title('Edge Detected (Canny)')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```
